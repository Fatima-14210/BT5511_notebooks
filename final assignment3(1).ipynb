```python
import os 
import pandas as pd 
```

```python
folder_path = "C:/Users/ojaim/Downloads/data_files"
all_data= []

for file in os.listdir(folder_path):
        if file.endswith(".txt"):
            df = pd.read_csv(os.path.join(folder_path,file),sep="\t")
            df['gene_file'] = file
            all_data.append(df)

full_df = pd.concat(all_data, ignore_index=True)
full_df
```

```python
from Bio import pairwise2 
def classify_mutation (wildtype_seq, mutant_seq):
    alignments = pairwise2.align.globalxx(wildtype_seq, mutant_seq, one_alignment_only=True)
    aligned_wildtype, aligned_mutant = alignments[0][:2]
    if len(wildtype_seq) > len(mutant_seq):
        return "Deletion"
    elif len(wildtype_seq) < len(mutant_seq):
        return "insertion"
    elif wildtype_seq != mutant_seq:
        return "substitution" 
    else:
        return "No change"
        
full_df["MutationType"] = full_df.apply(lambda row: classify_mutation(row["WildType.Sequence"], row["Mutant.Sequence"]), axis=1)
full_df
```

```python
#calculate cell viability for the wild type of each gean.
full_df ["WT_CellViability_Mean"] =full_df[["CellViability.WT.Rep1","CellViability.WT.Rep2","CellViability.WT.Rep3"]].mean(axis=1)
# calculate cell viability for mutant for each gene.
full_df ["Mut_CellViability_Mean"] = full_df[["CellViability.Mut.Rep1","CellViability.Mut.Rep2","CellViability.Mut.Rep3"]].mean(axis=1)
# calculate the difference between the cell viability mean of mutant and wild type 
full_df ["Delta_CellViability"]= ( full_df["Mut_CellViability_Mean"] - full_df["WT_CellViability_Mean"] )
print (full_df[["Gene", "WT_CellViability_Mean", "Mut_CellViability_Mean", "Delta_CellViability"]] )
full_df[["Gene", "WT_CellViability_Mean", "Mut_CellViability_Mean", "Delta_CellViability"]].to_csv("cell_viability_summary.csv", index=False)
```

```python
higher_cellviability = full_df.sort_values (by="Delta_CellViability", ascending=False)
print (higher_cellviability[["Gene", "WT_CellViability_Mean", "Mut_CellViability_Mean", "Delta_CellViability","MutationType"]] )
```

```python
import os 
import pandas as pd 
#calculate protein expression for the wild type of each geane.
full_df ["WT_Protein_Mean"] = full_df[["Protein.Expression.WT.Rep1","Protein.Expression.WT.Rep2","Protein.Expression.WT.Rep3"]].mean(axis=1)
# calculate protein expression for mutant for each gene.
full_df ["Mut_Protein_Mean"] = full_df[["Protein.Expression.Mut.Rep1","Protein.Expression.Mut.Rep2","Protein.Expression.Mut.Rep3"]].mean(axis=1)
# calculate the difference between the protein expression mean of mutant and wild type 
full_df ["Delta_Protein_Expression"]= ( full_df["Mut_Protein_Mean"] - full_df["WT_Protein_Mean"] )
print (full_df[["Gene", "WT_Protein_Mean", "Mut_Protein_Mean", "Delta_Protein_Expression"]] )
```

```python
higher_Protein_Expression = full_df.sort_values (by="Delta_Protein_Expression", ascending=False)
print (higher_Protein_Expression[["Gene", "WT_Protein_Mean", "Mut_Protein_Mean", "Delta_Protein_Expression", "MutationType"]] )
```

```python
def mutation_in_promoter (wt_seq, mut_seq, promoter_length=1000):
    for i in range (min(len(wt_seq), len(mut_seq))):
        if wt_seq[i] != mut_seq[i]:
             return i < promoter_length
    return False 
full_df ["MutationInPromoter"] = full_df.apply (lambda row: mutation_in_promoter(row["WildType.Sequence"], row["Mutant.Sequence"]), axis=1)
print (full_df [["Gene", "MutationInPromoter", "MutationType"]])                                                                                
```

```python
#calculate mRNA expression for the wild type of each geane.
full_df ["WT_mRNA_Mean"] = full_df[["mRNA.Expression.WT.Rep1","mRNA.Expression.WT.Rep2","mRNA.Expression.WT.Rep3"]].mean(axis=1)
# calculate protein expression for mutant for each gene.
full_df ["Mut_mRNA_Mean"] = full_df[["mRNA.Expression.Mut.Rep1","mRNA.Expression.Mut.Rep2","mRNA.Expression.Mut.Rep3"]].mean(axis=1)
# calculate the difference between the protein expression mean of mutant and wild type 
full_df ["Delta_mRNA_Expression"]= ( full_df["Mut_mRNA_Mean"] - full_df["WT_mRNA_Mean"] )
print (full_df[["Gene", "WT_mRNA_Mean", "Mut_mRNA_Mean", "Delta_mRNA_Expression"]] )
```

```python
higher_mRNA_Expression = full_df.sort_values (by="Delta_mRNA_Expression", ascending=False)
print (higher_mRNA_Expression[["Gene", "WT_mRNA_Mean", "Mut_mRNA_Mean", "Delta_mRNA_Expression", "MutationType"]])
```

```python
full_df ["result"]= full_df ["Delta_CellViability"] + full_df["Delta_Protein_Expression"] + full_df["Delta_mRNA_Expression"]
Top_5 = full_df.sort_values(by="result", ascending=False).head(5) 
print (Top_5[["Gene","Mut_CellViability_Mean","Mut_Protein_Mean","Mut_mRNA_Mean","Delta_CellViability","Delta_Protein_Expression","Delta_mRNA_Expression","MutationType"]])
```
